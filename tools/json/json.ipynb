# JSON Types 
1. `Number`	 Double precision floating-point
2. `String`	 Series of characters
3. `Boolean` true or false
4. `Array`	 Ordered sequence of values
5. `Value`	 String, Number, Boolean, null etc
6. `Object`	 Unordered collection of key/value pairs
7. `null`	 Null or Empty

# Numbers 
1. `Integer`	Digits 1-9, 0 and positive or negative
2. `Fraction`	Fractions like 0.3, 3.9
3. `Exponent`	Exponent like e, e+, e-, E, E+, E

```json
{
  "positive" : 12,
  "negative" : -1,
  "fraction" : 10.25,
  "exponent" : 1.0E+2,
  "zero" : 0
}
```

# Arrays 
* Begins with `[` and ends with `]`
* `[1, 2, 3, 4, 5]`

# Directory / Objects 
* Multiple key/value pairs separated by a comma. 
```json
{
  "color": "Purple",
  "id": "210",
  "composition": {
    "R": 70,
    "G": 39,
    "B": 89
  },
  "empty_object": {}
}
```

# Array of Objects 
```json
{
  "children": [
    {"name": "Jimmy Smith", "age": 15},
    {"name": "Sammy Sosa", "age": 12}
  ]
}
```

# Objects of Arrays 
```json
{
  "attributes": ["a1", "a2"],
  "methods": ["getter", "setter"],
  "empty_array": []
}
```

# 2D Array 
```json
{
  "my_sequences": [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9, 0],
    [10, 11]
  ]
}
```

# Object of Objects 
```json 
{
  "Mark McGwire": {
    "hr": 65,
    "avg": 0.278
  },
  "Sammy Sosa": {
    "hr": 63,
    "avg": 0.288
  }
}
```

# Nested 
```json
{
  "Jack": {
    "id": 1,
    "name": "Franc",
    "salary": 25000,
    "hobby": ["a", "b"],
    "location": {
        "country": "A", "city": "A-A"
    }
  }
}
```

# JQUERY or JPATH


- macOS: `brew install jq`

- Linux (Debian/Ubuntu): `sudo apt-get install jq`

- Windows: `choco install jq`

1. Pretty-Print: `jq . sample.json`
2. Get a Field: `jq '.name' sample.json` If you want Raw Strings without quotes use `-r` `jq -r '.name' sameple.json`
3. To get the Data in Brackets use Brackets: `jq '[.name]' sample.json`
3. Nested Fields: `jq '.user.email' sample.json`
4. Array Index: `jq '.[0]' sample.json`
5. Array Multiple Index: `jq '.[0,3]' data.json`
6. Slice Array: `jq '.[1:4]' data.json`
7. Step Slice: `jq '.[::2]' data.json`

# JQ Wild Cards
* To print all array Elements: `jq '.[]' file.json`
* All object values: `jq '.[]' file.json`, for a particular key we want to get a certain key for all its objects, then `jq '.[].color' sample.json`

* Print Kind: `jq '[.kind]' input.json`
* Print Pod Name: `jq '[.metadata.name]' input.json`
* Print Node Name: `jq '[.spec.nodeName]' input.json`
* Print Container Information: `jq '[.spec.containers[]]' input.json`
* Only print container Image Name: `cat input.json | jq '[.spec.containers[].image]'`
